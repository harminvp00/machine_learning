```python

import numpy as np 
features = [
    [2, 4, 6],
    [3, 5, 7],    
]

error = [-10, -20, -30]

gradients = np.zeros(3)

gradients[0] = np.sum(error) / 3;
for _f in range(2):
    gradients[_f + 1] = np.sum(np.multiply(features[_f], error)) / 3;

    
print("Gradient 0: ", gradients[0])
print("Gradient 1: ", gradients[1])
print("Gradient 2: ", gradients[2])
```
